## Experimento de rendimiento

El experimento de rendimiento se ejecuta mediante el script *scripts/AL_performance.py*, que recibe los siguientes argumentos de entrada:
- **Número de cores**. Número de cores utilizados para configurar la *SparkSession* en modo local. En este experimento se configuran 4 cores.
- **Tamaño del dataset**. Porcentaje del dataset de referencia utilizado. Para el experimento se usa el $40\%$.
- **Método de selección.** Estrategia empleada para seleccionar las instancias que serán etiquetadas. Se ejecutan dos configuraciones: `smart`, correspondiente a la estrategia propuesta basada en incertidumbre y diversidad, y `random`, correspondiente a la selección aleatoria utilizada como referencia.
- **Número de iteraciones.** Número de ciclos de *Active Learning* ejecutados. Cada iteración comprende el entrenamiento y evaluación del modelo, la selección de instancias y la actualización de sus etiquetas. En este experimento se ejecutan 10 iteraciones.
- **Fichero de resultados.** Nombre del fichero en el que se almacenan las métricas obtenidas durante la ejecución.

```python
# Experiment configuration
num_cores = 4
dataset_size = 40
AL_iterations = 10
output_filename = "performance.csv"
```
