## Performance Experiment

The performance experiment is executed using the *scripts/AL_performance.py* script, which accepts the following input arguments:

* **Number of cores.** Number of cores used to configure the *SparkSession* in local mode. In this experiment, $4$ cores are configured.

* **Dataset size.** Percentage of the reference dataset used. The experiment uses $40%$ of the reference dataset.

* **Selection method.** Strategy used to select the instances to be labeled. Two configurations are evaluated: `smart`, corresponding to the proposed uncertainty- and diversity-based strategy, and `random`, corresponding to the random selection used as a reference.

* **Number of iterations.** Number of *Active Learning* cycles executed. Each iteration consists of model training and evaluation, instance selection, and label state updating. A total of $10$ iterations are performed in this experiment.

* **Results file.** Name of the file in which the metrics obtained during the execution are stored.

```python
# Experiment configuration
num_cores = 4
dataset_size = 40
AL_iterations = 10
output_filename = "performance.csv"
```

```python
# Method selection = smart
method_selection = "smart"
%run scripts/AL_performance.py $num_cores $dataset_size $method_selection $AL_iterations $output_filename
```

```python
# Method selection = random
method_selection = "random"
%run scripts/AL_performance.py $num_cores $dataset_size $method_selection $AL_iterations $output_filename
```
